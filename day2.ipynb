```python
# accessing attribute and methods
class emp:
    name='maanya'
    salary='25000'
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1=emp()
```

```python
# using getattr insted e1.name
print(getattr(e1,'name'))
```

```python
# hasattr - return true if the object has attribute
print(hasattr(e1,'name'))
```

```python
print(hasattr(e1,'gender'))
```

```python
# setattr -sets an attribute
setattr(e1,'gender','f')
```

```python
print(getattr(e1,'gender'))
```

```python
# delattr-del the attribute
delattr(e1,'gender')
```

```python
#it should show attribute erro since it is deleted
print(getattr(e1,'gender'))
```

```python
class student:
    def __init__(self,usn,name,age): #constructor
        #instance variable
        self.usn=usn
        self.name=name
        self.age=age
        #instance method access instance variable
    def show(self):
            print('usn:',self.usn,',name:',self.name,',age:',self.age)
            # instance method to  modify instance variable
    def update(self,usn,age):
                self.usn=usn
                self.age=age
        
```

```python
stu=student(153,'maanya',21)
stu.show()
```
