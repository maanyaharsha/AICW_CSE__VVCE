```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    @classmethod
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year-birthyear)

    def show(self):
        print(self.name+"'s age is '"+str(self.age))
```

```python
age=student('maanya',21)
age.show()
```

```python
by=student.calculate_age('maanya',2005)
by.show()
```

```python
# mini project
class BankAccount:
    def __init__(self,name,balance=0):
        #instance attributes
        self.name=name
        self.balance=balance
    # deposite money
    def deposite(self,amount):
        if amount>0:
            self.balance+=amount
            print("deposite sucessfully!")
            print("deposited:",amount)
        else:
            print("invalid deposite amount")
    #withdraw money
    def withdraw(self,amount):
        if amount <=0:
            print('invalid withdrawl amount!')
        elif amount>self.balance:
            print('insufficient balance!')
        else:
            self.balance-=amount
            print("withdrawal sucessfully!")
            print("withdrawn :",amount)
        #check balance
    def check_balance(self):
        print("current balance:",self.balance)
    #display the account details
    def display_details(self):
        print("account holder:",self.name)
        print("balance:",self.balance)
```

```python
acc1=BankAccount("maanya",1000)
```

```python
acc1.display_details()
```

```python
acc1.deposite(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(2000)
```

```python
acc1.check_balance()
```

```python
# parent class
class Person:
    def person_info(self,name,age):
        print('inside person class')
        print('name:',name,'age:',age)
#parent class 2
class Company:
    def company_info(self,company_name,location):
        print('inside the company class')
        print('name:',company_name,'location:',location)
        
```

```python
#child class
class Employee(Person,Company):
    def Employee_info(self,salary,skill):
        print('inside employee class-child class')
        print('salary',salary,'skills:',skill)
```

```python
#create an object of employee
emp2=Employee()
```

```python
#acess the data
emp2.person_info('maanya',21)
emp2.company_info('edunet foundation','bengaluru')
emp2.Employee_info(25000,'aiml engineer')
```

```python
#hybrid inheritance
class Vehicle:
    def vehicle_info(self):
        print('Inside Vehicle class')

class Car(Vehicle):
    def car_info(self):
        print('Inside Car class')

class Truck(Vehicle):
    def truck_info(self):
        print('Inside Truck class')

class SportsCar(Car):
    def sports_car_info(self):
        print('Inside SportsCar class')

# Create object
s = SportsCar()

s.vehicle_info()
s.car_info()
s.sports_car_info()

t = Truck()
t.vehicle_info()
t.truck_info()
```

```python
# public member
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    #public instance method
    def show(self):
        #access public datamembers
        print('name',self.name,'salary',self.salary)
```

```python
# protected member
class Company:
    def __init__(self):
        #protected member
        self._project="nlp"
#child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("employee Name:",self.name)
        #Access the protected member
        print("working on project:",self._project)
```

```python
c=Employee("maanya")
c.show()
```

```python
# direct acess
print('project:',c._project)
```

```python
#over loading
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a+other.a,self.b+other.b
ob1=complex(1,2)
ob2=complex(2,3)
ob3=ob1+ob2
print(ob3)
```
